```python

def simple_chatbot():
    print("Hello! I'm a simple chatbot. Type 'quit' to exit.")

    while True:
        user_input = input("You: ")

        if user_input.lower() == 'quit':
            print("Chatbot: Goodbye!")
            break
        elif "hello" in user_input.lower() or "hi" in user_input.lower():
            print("Chatbot: Hi there! How can I help you today?")
        elif "how are you" in user_input.lower():
            print("Chatbot: I'm just a program, but I'm functioning well! How about you?")
        elif "name" in user_input.lower():
            print("Chatbot: I don't have a name. You can call me Chatbot.")
        elif "weather" in user_input.lower():
            print("Chatbot: I cannot tell you the weather, as I don't have access to real-time information.")
        else:
            print(f"Chatbot: You said: '{user_input}'. I'm not sure how to respond to that.")

simple_chatbot()
```
